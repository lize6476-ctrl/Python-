# University Recommendation System
A program that recommends a profession based on your interests and suggests the best universities where it is taught.

```python
professions = {
    "Computer Science": {
        "keywords": ["programming", "math", "computers", "technology", "games", "data", "algorithms"],
        "universities": [
            ("MIT", "USA", 99),
            ("Stanford University", "USA", 98),
            ("Carnegie Mellon University", "USA", 97),
            ("ETH Zurich", "Switzerland", 96),
            ("University of Oxford", "UK", 95),
        ]
    },
    "Medicine": {
        "keywords": ["biology", "chemistry", "health", "science", "anatomy", "nature", "doctor"],
        "universities": [
            ("Harvard Medical School", "USA", 99),
            ("Johns Hopkins University", "USA", 98),
            ("University of Oxford", "UK", 97),
            ("University of Cambridge", "UK", 96),
            ("Karolinska Institute", "Sweden", 95),
        ]
    },
    "Business & Economics": {
        "keywords": ["business", "money", "finance", "economics", "management", "marketing", "trade"],
        "universities": [
            ("Harvard Business School", "USA", 99),
            ("Wharton School (UPenn)", "USA", 98),
            ("London Business School", "UK", 97),
            ("INSEAD", "France", 96),
            ("MIT Sloan", "USA", 95),
        ]
    },
    "Engineering": {
        "keywords": ["physics", "mechanics", "electronics", "robots", "energy", "machines", "construction"],
        "universities": [
            ("MIT", "USA", 99),
            ("Caltech", "USA", 98),
            ("ETH Zurich", "Switzerland", 97),
            ("Imperial College London", "UK", 96),
            ("TU Munich", "Germany", 95),
        ]
    },
    "Psychology": {
        "keywords": ["people", "behavior", "brain", "emotions", "communication", "therapy", "society"],
        "universities": [
            ("Harvard University", "USA", 99),
            ("Stanford University", "USA", 98),
            ("University of Cambridge", "UK", 97),
            ("UCLA", "USA", 96),
            ("University of Amsterdam", "Netherlands", 95),
        ]
    },
    "Law": {
        "keywords": ["law", "justice", "politics", "rights", "history", "government", "society"],
        "universities": [
            ("Harvard Law School", "USA", 99),
            ("Yale Law School", "USA", 98),
            ("University of Oxford", "UK", 97),
            ("University of Cambridge", "UK", 96),
            ("NYU School of Law", "USA", 95),
        ]
    },
}

fields = list(professions.keys())
countries = set(u[1] for p in professions.values() for u in p["universities"])

print(f"Fields available: {len(fields)}")
print(f"Countries represented: {len(countries)} — {', '.join(sorted(countries))}")
```

```python
def match_score(interests, keywords):
    user_set = set(interests)
    kw_set = set(keywords)
    score = len(user_set & kw_set) * 10
    for i in interests:
        for k in keywords:
            if (i in k or k in i) and i not in (user_set & kw_set):
                score += 5
                break
    return score


def find_professions(interests):
    scores = {f: match_score(interests, professions[f]["keywords"]) for f in fields}
    results = [(f, s) for f, s in scores.items() if s > 0]
    return sorted(results, key=lambda x: x[1], reverse=True)


def show_universities(field):
    unis = sorted(professions[field]["universities"], key=lambda u: u[2], reverse=True)
    print(f"\nTop Universities for {field}:\n")
    for i in range(len(unis)):
        name, country, rating = unis[i]
        print(f"  {i + 1}. {name} — {country}  (Rating: {rating}/100)")


def show_menu():
    print("\n" + "=" * 50)
    print("    UNIVERSITY RECOMMENDATION SYSTEM")
    print("=" * 50)
    print("  1. Find profession by interests")
    print("  2. Browse universities by field")
    print("  3. List all fields")
    print("  4. Exit")
    print("=" * 50)
```

```python
running = True

while running:
    show_menu()
    choice = input("\nEnter choice (1-4): ").strip()

    if choice == "1":
        raw = input("Enter your interests (comma-separated): ").strip()
        interests = [x.strip().lower() for x in raw.split(",") if x.strip()]

        if not interests:
            print("No interests entered.")
        else:
            results = find_professions(interests)
            print(f"\nYour interests: {', '.join(interests)}\n")

            if results:
                print("Matching professions:\n")
                for rank, (field, score) in enumerate(results, 1):
                    if score >= 20:
                        level = "Strong match"
                    elif score >= 10:
                        level = "Good match"
                    else:
                        level = "Partial match"
                    print(f"  {rank}. {field} — {level} (score: {score})")

                see = input(f"\nSee universities for '{results[0][0]}'? (yes/no): ").strip().lower()
                if see in ["yes", "y"]:
                    show_universities(results[0][0])
            else:
                print("No matches found. Try different keywords.")

    elif choice == "2":
        print("\nAvailable fields:\n")
        for i, field in enumerate(fields, 1):
            print(f"  {i}. {field}")
        pick = input("\nEnter field number: ").strip()
        if pick.isdigit() and 1 <= int(pick) <= len(fields):
            show_universities(fields[int(pick) - 1])
        else:
            print("Invalid number.")

    elif choice == "3":
        print("\nAll fields:\n")
        for i, field in enumerate(fields, 1):
            print(f"  {i}. {field}")

    elif choice == "4":
        print("\nGoodbye!")
        running = False

    else:
        print("Invalid choice. Please enter 1, 2, 3, or 4.")
```
