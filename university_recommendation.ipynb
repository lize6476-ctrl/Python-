```python
# ============================================================
# БЛОК 2: ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# Используется: функции, str методы, list методы, if/elif/else
# ============================================================

def print_separator(char="=", length=60):
    """Печатает разделительную линию"""
    print(char * length)


def print_header(title):
    """Печатает заголовок с рамкой"""
    print_separator()
    # Метод строки: .center() для центрирования
    print(title.center(60))
    print_separator()


def get_valid_input(prompt, valid_options):
    """
    Получает корректный ввод от пользователя.
    Используется: while цикл, if/else, str методы (.strip(), .lower())
    """
    while True:
        user_input = input(prompt).strip()  # str метод: .strip()
        if user_input.lower() in valid_options:  # str метод: .lower()
            return user_input.lower()
        else:
            print(f"  ❌ Неверный ввод. Допустимые варианты: {', '.join(valid_options)}")


def format_university_info(rank, uni_tuple):
    """
    Форматирует информацию об университете для вывода.
    Используется: tuple (распаковка), str методы, f-строки
    """
    # Распаковка кортежа
    name, country, rating = uni_tuple
    stars = "⭐" * (rating // 20)  # str метод: умножение строки
    return f"  {rank}. {name}\n     🌍 Страна: {country} | Рейтинг: {rating}/100 {stars}"


def display_all_professions():
    """
    Выводит список всех профессий.
    Используется: enumerate(), for цикл, список
    """
    print("\n📚 Доступные направления:")
    print_separator("-")
    # enumerate даёт индекс и значение
    for i, profession in enumerate(all_professions, start=1):
        emoji = professions_data[profession]["emoji"]
        print(f"  {i}. {emoji} {profession}")
    print_separator("-")


def get_interests_from_user():
    """
    Запрашивает интересы пользователя.
    Используется: while цикл, list методы (.append(), .split()), str методы
    """
    print("\n💡 Введите ваши интересы (через запятую).")
    print("   Например: математика, компьютеры, игры")
    print("   Или одно слово: биология")
    
    while True:
        raw_input = input("\n✏️  Ваши интересы: ").strip()
        
        if raw_input == "":
            print("  ❌ Поле не может быть пустым. Попробуйте ещё раз.")
            continue
        
        # str метод: .split() разбивает строку по разделителю
        interests_list = raw_input.split(",")
        
        # list метод: .append() и list comprehension с str методами
        cleaned_interests = []
        for interest in interests_list:
            cleaned = interest.strip().lower()  # str методы: .strip(), .lower()
            if cleaned != "":
                cleaned_interests.append(cleaned)  # list метод: .append()
        
        if len(cleaned_interests) == 0:
            print("  ❌ Не удалось распознать интересы. Попробуйте ещё раз.")
        else:
            return cleaned_interests


print("✅ Вспомогательные функции загружены!")
```

```python
# ============================================================
# БЛОК 3: ФУНКЦИИ РЕКОМЕНДАЦИЙ
# Используется: функции, dict, list, set, sorted(), for, if/elif
# ============================================================

def calculate_match_score(user_interests, profession_keywords):
    """
    Вычисляет степень совпадения интересов с профессией.
    Используется: set (пересечение множеств), list, for цикл
    """
    # Преобразуем в множество для эффективного поиска
    interests_set = set(user_interests)
    keywords_set = set(profession_keywords)
    
    # Прямые совпадения: пересечение множеств
    direct_matches = interests_set & keywords_set  # оператор & для set
    score = len(direct_matches) * 10
    
    # Частичные совпадения: ищем вхождение подстрок
    for interest in user_interests:
        for keyword in profession_keywords:
            # str метод: .startswith() или проверка вхождения
            if interest in keyword or keyword in interest:
                if interest not in direct_matches:  # не считать дважды
                    score += 5
                    break
    
    return score


def recommend_professions(user_interests):
    """
    Находит подходящие профессии по интересам.
    Используется: dict, list, sorted(), for цикл, if условие
    """
    # Словарь для хранения баллов по каждой профессии
    scores = {}
    
    for profession, data in professions_data.items():
        score = calculate_match_score(user_interests, data["keywords"])
        if score > 0:
            scores[profession] = score
    
    if len(scores) == 0:
        return []  # пустой список если нет совпадений
    
    # sorted() с key — сортировка по убыванию баллов
    sorted_professions = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    
    return sorted_professions


def recommend_universities(profession_name, top_n=5, filter_country=None):
    """
    Рекомендует университеты для выбранной профессии.
    Используется: dict, list, tuple, sorted(), for, if/elif, range
    """
    if profession_name not in professions_data:
        return []
    
    universities = professions_data[profession_name]["universities"]
    
    # Фильтрация по стране, если указана
    if filter_country is not None:
        universities = [u for u in universities if u[1].lower() == filter_country.lower()]
    
    # Сортировка по рейтингу (3й элемент кортежа) — sorted()
    sorted_unis = sorted(universities, key=lambda u: u[2], reverse=True)
    
    # Используем range для ограничения количества
    result = []
    for i in range(min(top_n, len(sorted_unis))):
        result.append(sorted_unis[i])
    
    return result


def display_recommendations(sorted_professions, user_interests):
    """
    Выводит результаты рекомендаций профессий.
    Используется: for цикл, if/elif/else, enumerate(), list методы
    """
    print_header("  🎯 РЕЗУЛЬТАТЫ АНАЛИЗА ИНТЕРЕСОВ  ")
    
    # str метод: .join()
    print(f"\n🔍 Ваши интересы: {', '.join(user_interests)}")
    print()
    
    if len(sorted_professions) == 0:
        print("😔 По вашим интересам не найдено подходящих профессий.")
        print("   Попробуйте другие ключевые слова.")
        return
    
    print("📊 Подходящие профессии (по степени совпадения):")
    print_separator("-")
    
    for rank, (profession, score) in enumerate(sorted_professions, start=1):
        emoji = professions_data[profession]["emoji"]
        description = professions_data[profession]["description"]
        
        # if/elif/else для определения уровня совпадения
        if score >= 30:
            match_label = "Отличное совпадение ✨"
        elif score >= 20:
            match_label = "Хорошее совпадение 👍"
        elif score >= 10:
            match_label = "Частичное совпадение 🔶"
        else:
            match_label = "Слабое совпадение"
        
        print(f"  {rank}. {emoji} {profession}")
        print(f"     📝 {description}")
        print(f"     🎯 {match_label} (баллы: {score})")
        print()


def display_universities_for_profession(profession_name):
    """
    Выводит топ университетов для профессии.
    Используется: функции, for цикл, tuple, enumerate()
    """
    emoji = professions_data[profession_name]["emoji"]
    print_header(f"  {emoji} Топ Университетов  ")
    print(f"\n  Направление: {profession_name}")
    print_separator("-")
    
    universities = recommend_universities(profession_name)
    
    if len(universities) == 0:
        print("  Университеты не найдены.")
        return
    
    for rank, uni_tuple in enumerate(universities, start=1):
        print(format_university_info(rank, uni_tuple))
        print()


print("✅ Функции рекомендаций загружены!")
```

```python
# ============================================================
# БЛОК 4: ГЛАВНОЕ МЕНЮ ПРОГРАММЫ
# Используется: while цикл, if/elif/else, функции, input()
# ============================================================

def show_main_menu():
    """Выводит главное меню программы"""
    print_header("  🎓 СИСТЕМА ВЫБОРА УНИВЕРСИТЕТА  ")
    print()
    print("  Выберите действие:")
    print()
    print("  1. 🔍 Найти профессию по интересам")
    print("  2. 🏛️  Просмотреть университеты по направлению")
    print("  3. 📋 Список всех направлений")
    print("  4. ℹ️  О программе")
    print("  5. 🚪 Выход")
    print()
    print_separator("-")


def handle_interest_search():
    """
    Обрабатывает поиск профессии по интересам.
    Используется: функции, while, if/elif, list, dict
    """
    print_header("  🔍 ПОИСК ПО ИНТЕРЕСАМ  ")
    
    user_interests = get_interests_from_user()
    results = recommend_professions(user_interests)
    display_recommendations(results, user_interests)
    
    if len(results) == 0:
        return
    
    # Предлагаем посмотреть университеты для лучшей профессии
    best_profession = results[0][0]  # первый элемент — название профессии
    
    print_separator("-")
    print(f"\n💡 Лучшее совпадение: {best_profession}")
    print("   Хотите увидеть топ университетов для этого направления?")
    
    choice = get_valid_input("   Ваш выбор (да/нет): ", ["да", "нет", "д", "н"])
    
    if choice in ["да", "д"]:
        display_universities_for_profession(best_profession)
        
        # Если результатов больше одного — предлагаем посмотреть другие
        if len(results) > 1:
            print_separator("-")
            print("\n📌 Другие подходящие направления:")
            for i, (prof, score) in enumerate(results[1:4], start=2):  # slice list
                emoji = professions_data[prof]["emoji"]
                print(f"   {i}. {emoji} {prof} (баллы: {score})")


def handle_browse_universities():
    """
    Позволяет выбрать направление и просмотреть университеты.
    Используется: while, for, range, list, dict, if/elif/else
    """
    display_all_professions()
    
    # Создаём список допустимых номеров как строки
    valid_numbers = [str(i) for i in range(1, len(all_professions) + 1)]
    
    while True:
        choice_str = input("\n  Введите номер направления (или 0 для отмены): ").strip()
        
        if choice_str == "0":
            print("  ↩️  Возврат в главное меню.")
            return
        elif choice_str in valid_numbers:
            # Преобразование в индекс (списки с 0)
            index = int(choice_str) - 1
            selected = all_professions[index]  # обращение к элементу списка
            break
        else:
            print(f"  ❌ Введите число от 1 до {len(all_professions)} или 0 для отмены.")
    
    display_universities_for_profession(selected)


def show_about():
    """Показывает информацию о программе"""
    print_header("  ℹ️  О ПРОГРАММЕ  ")
    print()
    info_lines = [
        "  Программа помогает выбрать направление обучения",
        "  и найти лучшие университеты мира.",
        "",
        "  Как работает система:",
        "  • Вы вводите свои интересы",
        "  • Система анализирует совпадение с профессиями",
        "  • Показывает подходящие направления и университеты",
        "",
        f"  Направлений в базе: {len(all_professions)}",
        f"  Стран с университетами: {len(all_countries)}",
    ]
    # list метод: for по списку строк
    for line in info_lines:
        print(line)
    print()


print("✅ Функции меню загружены!")
```

```python
# ============================================================
# БЛОК 5: ЗАПУСК ПРОГРАММЫ
# Используется: while цикл, if/elif/else, все предыдущие функции
# ============================================================

def run_program():
    """
    Главная функция — запускает программу.
    Используется: while цикл, if/elif/else, все изученные темы
    """
    print("\n" + "=" * 60)
    print("  Добро пожаловать!".center(60))
    print("=" * 60)
    
    # Флаг для управления циклом
    running = True
    
    # Основной цикл программы (while)
    while running:
        print()
        show_main_menu()
        
        # Получаем выбор пользователя
        user_choice = input("  Ваш выбор (1-5): ").strip()
        print()
        
        # if/elif/else для обработки меню
        if user_choice == "1":
            handle_interest_search()
        
        elif user_choice == "2":
            handle_browse_universities()
        
        elif user_choice == "3":
            display_all_professions()
        
        elif user_choice == "4":
            show_about()
        
        elif user_choice == "5":
            print("  👋 До свидания! Удачи в выборе университета!")
            print_separator()
            running = False  # завершаем цикл
        
        else:
            print("  ❌ Неверный выбор. Введите число от 1 до 5.")
        
        # Пауза между действиями (кроме выхода)
        if running:
            input("\n  Нажмите Enter для продолжения...")


# Запускаем программу!
run_program()
```

## 📝 Краткое описание использованных тем

| Тема | Где используется |
|------|------------------|
| `if / elif / else` | Обработка меню, уровни совпадения, проверка ввода |
| `while` | Главный цикл программы, циклы валидации ввода |
| `list`, `for`, `range` | Хранение профессий, перебор данных, ограничение топа |
| `tuple` | Данные университетов (название, страна, рейтинг) |
| Методы `str` | `.strip()`, `.lower()`, `.split()`, `.join()`, `.center()` |
| Методы `list` | `.append()`, индексация, срезы |
| `set` | Множество стран, пересечение для подсчёта совпадений |
| `dict` | База данных профессий и университетов |
| Вложенные структуры | `dict` содержит `list` кортежей |
| `sorted()` | Сортировка профессий по баллам, университетов по рейтингу |
| Функции | Каждый логический блок — отдельная функция |
